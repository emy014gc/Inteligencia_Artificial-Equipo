```python
import tensorflow as tf
from tensorflow.keras import datasets, models, layers
import numpy as np
import matplotlib.pyplot as plt # Opcional, para visualizar los números
```

```python
# Carga de datos directamente desde Keras
(train_data, train_labels), (test_data, test_labels) = datasets.mnist.load_data()

print(f"Forma de los datos de entrenamiento: {train_data.shape}")
print(f"Forma de los datos de prueba: {test_data.shape}")
```

```python
# Normalización de los datos (escalar los valores entre 0 y 1)
train_data = train_data / 255.0
test_data = test_data / 255.0
```

```python
# 1. Definir la estructura del modelo
model = models.Sequential([
    layers.Flatten(input_shape=(28, 28)),          # Capa de entrada: aplana la matriz 28x28 a un vector de 784
    layers.Dense(128, activation='relu'),          # Capa oculta con 128 neuronas y activación ReLU
    layers.Dense(10, activation='softmax')         # Capa de salida con 10 neuronas (dígitos 0-9) y activación softmax
])

# 2. Compilar el modelo (establecer optimizador, pérdida y métrica)
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# 3. Entrenar el modelo con los datos de entrenamiento
print("Iniciando el entrenamiento...")
history = model.fit(train_data, train_labels, epochs=5)
```

```python
# Evaluar la exactitud del modelo con test_data
print("\n--- Evaluación del Modelo ---")
test_loss, test_acc = model.evaluate(test_data, test_labels, verbose=2)
print(f"Exactitud (Accuracy) en el conjunto de prueba: {test_acc:.4f}")

# Realizar 5 predicciones diferentes
print("\n--- Resultados de 5 Predicciones ---")
predicciones = model.predict(test_data[:5])

for i in range(5):
    # np.argmax obtiene el índice con la probabilidad más alta (el dígito predicho)
    prediccion_red = np.argmax(predicciones[i])
    valor_real = test_labels[i]

    print(f"Imagen {i+1} -> Predicción: {prediccion_red} | Valor Real: {valor_real}")
```
