```python
import tensorflow as tf
from tensorflow import keras
import numpy as np
```

```python
# Cargar datos MNIST
(train_data, train_labels), (test_data, test_labels) = keras.datasets.mnist.load_data()

print(f"Forma original de los datos de entrenamiento: {train_data.shape}")
print(f"Forma original de los datos de prueba: {test_data.shape}")
```

```python
# Normalizar valores de píxeles (escalar los valores entre 0 y 1)
train_data = train_data.astype('float32') / 255.0
test_data = test_data.astype('float32') / 255.0

# Redimensionar para CNN -> (28,28,1)
train_data = train_data.reshape((60000, 28, 28, 1))
test_data = test_data.reshape((10000, 28, 28, 1))

# Convertir etiquetas a one-hot encoding
train_labels_cat = keras.utils.to_categorical(train_labels, 10)
test_labels_cat = keras.utils.to_categorical(test_labels, 10)
```

```python
# 1. Definir modelo CNN
model = keras.Sequential([
    # Primera capa convolucional
    keras.layers.Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)),
    keras.layers.MaxPooling2D((2,2)), # Pooling

    # Segunda capa convolucional
    keras.layers.Conv2D(64, (3,3), activation='relu'),
    keras.layers.MaxPooling2D((2,2)), # Segundo pooling

    # Aplanar datos
    keras.layers.Flatten(),

    # Capa oculta
    keras.layers.Dense(64, activation='relu'),

    # Capa de salida
    keras.layers.Dense(10, activation='softmax')
])

# 2. Compilar modelo
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# 3. Entrenar modelo
print("Iniciando el entrenamiento...")
history = model.fit(train_data, train_labels_cat,
                    epochs=10,
                    batch_size=128,
                    validation_data=(test_data, test_labels_cat),
                    verbose=1)
```

```python
# Evaluar modelo
test_loss, test_acc = model.evaluate(test_data, test_labels_cat, verbose=0)
print(f'\nExactitud en prueba: {test_acc:.4f}')

# Realizar 5 predicciones
print('\n--- 5 predicciones vs valores reales ---')
predicciones = model.predict(test_data[:5])

for i in range(5):
    pred = np.argmax(predicciones[i])
    real = test_labels[i]
    print(f'Predicción: {pred} | Valor real: {real}')
```
